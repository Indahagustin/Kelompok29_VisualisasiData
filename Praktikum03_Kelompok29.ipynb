Import

```python
import matplotlib.pyplot as plt
```

Dataset

```python
# Buat data sample
month = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Jun','Jul','Aug', 'Sep', 'Oct', 'Nov', 'Dec']
product_A_sales = [10,20,15,25,30,45,40,50,60,55,56,70]
product_B_sales = [5,10,8,15,18,20,22,30,25,35,40,45]
```

**Single Line Plot**

```python
#line plot 1 garis, style grafik default
plt.plot(month, product_A_sales, label="Product A")
plt.title('Penjualan Produk A per Bulan')
plt.xlabel('Bulan')
plt.ylabel('Penjualan')
plt.legend()
plt.grid(True)
plt.show()
```

**Multiple Line Plot & Customization**

```python
from matplotlib.lines import lineStyles
# kustomisasi line plot (line 1 - product A)
plt.plot(months, product_A_sales, label='Product A', linestyle="--", marker='o', color='blue')

# kustomisasi line plot (line 2 - product B)
plt.plot(months, product_B_sales, label='Product B', linestyle="-", marker='x', color='red')

plt.title('Penjualan Produk A per Bulan')
plt.xlabel('Bulan')
plt.ylabel('Penjualan')
plt.legend()
plt.grid(True)
plt.show()
```

**Jenis Garis untuk Menunjukkan Tren**

```python
# Data sample tambahan
product_C_sales = [8,12,14,18,20,25,28,30,32,35,38,40]
product_D_sales = [12,15,10,18,22,30,25,20,28,32,30,38]

# plot product A (dashed)
plt.plot(months, product_A_sales, label='Product A', linestyle="--", color='blue')

# plot product B (solid)
plt.plot(months, product_B_sales, label='Product B', linestyle="-", color='red')

# plot product C (dash-dot)
plt.plot(months, product_C_sales, label='Product c', linestyle=":", color='green')

# plot product D (dotted)
plt.plot(months, product_D_sales, label='Product D', linestyle="-.", color='purple')

plt.title('Penjualan Produk A per Bulan')
plt.xlabel('Bulan')
plt.ylabel('Jumlah Penjualan')
plt.legend()
plt.grid(True)
plt.show()
```

**Subplot pada Line Chart**

```python
# subplot, 2 baris 1 kolom, panjang 10 inch, tinggi 8 inch
fig, axs = plt.subplots(2, 1, figsize=(10, 8))

# plot pertama untuk product C
axs[0].plot(months, product_C_sales, label='Product C', linestyle=":", color='green', marker='d')
axs[0].set_title('Penjualan Product C per Bulan')
axs[0].set_xlabel('Bulan')
axs[0].set_ylabel('Jumlah Penjualan')
axs[0].legend()
axs[0].grid(True)

# plot pertama untuk product D
axs[1].plot(months, product_D_sales, label='Product D', linestyle=":", color='purple', marker='s')
axs[1].set_title('Penjualan Product D per Bulan')
axs[1].set_xlabel('Bulan')
axs[1].set_ylabel('Jumlah Penjualan')
axs[1].legend()
axs[1].grid(True)

plt.tight_layout()
plt.show()
```
